Skip-Gram模型训练

```python
# 词向量
# 单元格1：Skip-Gram模型训练
import pandas as pd
import jieba
from gensim.models import Word2Vec

# 数据预处理（假设已加载数据）
data = pd.read_csv('train.csv')
corpus = data['comment'].values.astype(str)
corpus = [jieba.lcut(text.replace("，", "").replace("!", "").replace("！", "")
                   .replace("。", "").replace("~", "").replace("；", "")
                   .replace("？", "").replace("?", "").replace("【", "")
                   .replace("】", "").replace("#", "")) for text in corpus]

# 使用Skip-Gram训练（sg=1）
model = Word2Vec(corpus, sg=1, vector_size=300, window=5, min_count=3, workers=4)
```

词向量输出

```python
# 单元格2：词向量输出
print("'环境'的词向量：\n", model.wv['环境'])
print("\n向量形状：", model.wv['环境'].shape)
```

语义相近词

```python
# 单元格3：语义相近词
print("最接近'好吃'的3个词：", model.wv.most_similar('好吃', topn=3))
```

相似度计算

```python
# 单元格4：相似度计算
print("好吃-美味相似度：", round(model.wv.similarity('好吃', '美味'), 4))
print("好吃-蟑螂相似度：", round(model.wv.similarity('好吃', '蟑螂'), 4))
```

向量运算

```python
# 单元格5：向量运算
result = model.wv.most_similar(positive=['餐厅', '聚会'], negative=['安静'], topn=1)
print("餐厅 + 聚会 - 安静 =", result[0][0])
```
